## 1. Load and clean PRODUCTLINE

```python
import pandas as pd

df = pd.read_csv("sales_data_sample.csv", encoding="latin1")

df["PRODUCTLINE"] = df["PRODUCTLINE"].str.strip().str.title()

print(df[["PRODUCTLINE"]].head())
```

## 2. Clean SALES and convert to float

```python
df["SALES"] = (
    df["SALES"]
    .astype(str)
    .str.replace(r"[^0-9.\-]", "", regex=True)
    .astype(float)
)

print(df["SALES"].head())
print("SALES data type:", df["SALES"].dtype)
```

## 3. Build cumulative PRODUCTLINE sales dictionary

```python
product_sales = {}

for product_line, sales in zip(df["PRODUCTLINE"], df["SALES"]):
    if product_line not in product_sales:
        product_sales[product_line] = 0

    product_sales[product_line] += sales

print(product_sales)
```

## 4. Categorize High Revenue Products

```python
for product_line, total_sales in product_sales.items():
    if total_sales > 100000:
        print(product_line, "-", "High Revenue Product", "-", total_sales)
```
