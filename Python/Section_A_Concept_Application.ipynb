## 1. String methods and execution order

```python
product_name = "  classic cars  "
cleaned_name = product_name.strip().title()
print(cleaned_name)
# Execution order: strip() removes surrounding spaces, then title() fixes casing.
```

## 2. Converting numeric strings to floats

```python
sales_value = "12345.50"
sales_value = float(sales_value)
print(sales_value)
print(type(sales_value))
# float() converts a numeric string into a floating-point number.
```

## 3. Mapping product names to total sales

```python
product_sales = {"Classic Cars": 120000.50, "Motorcycles": 85000.25}
print(product_sales)
# A dictionary maps each product name directly to its total sales.
# A list is less suitable because a product must be searched for manually.
```

## 4. Manual for loop versus sum()

```python
sales = [100, 250, 300]
print("Using sum():", sum(sales))

total = 0
for value in sales:
    if value > 200:
        total += value
print("Using loop with condition:", total)
# A loop is necessary when each value needs additional conditional processing.
```

## 5. Preventing TypeError

```python
value = "500"
try:
    value = float(value)
    if value > 400:
        print("Value is greater than 400")
    else:
        print("Value is 400 or less")
except (TypeError, ValueError):
    print("Invalid numeric value")
```
